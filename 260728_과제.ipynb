```python
def get_middle_char(i):
    mid_idx = len(i) // 2
    if len(i) % 2 == 0:                 # 조건 : 문자열의 길이가 짝수일 경우
        return i[mid_idx-1 : mid_idx+1]
    else:                               # 조건 : 문자열의 길이가 짝수가 아닐 경우(=홀수)
        return i[mid_idx]

print(get_middle_char('study'))
print(get_middle_char('python'))
print(get_middle_char('workspace'))
```

```python
def my_avg(*args):
    return sum(args) / len(args)   # 튜플 상태에서는 내장 함수를 바로 사용하여 구할 수 있음 
    
my_avg(77, 83, 95, 80, 70) #=> 81.0
```
