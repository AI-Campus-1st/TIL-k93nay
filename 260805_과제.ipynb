## 1. Numpy 배열로 5명 학생의 4과목 성적표를 만들고, 축(axis)기준 통계를 산출하시오.

아래 1차원 리스트는 학생 5명이 4개 과목(국어, 영어, 수학, 과학 순)에서 받은 점수를 이어 붙인 것입니다.

Numpy만 사용하여(반복문 사용 금지) 다음을 순서대로 수행하고 결과를 출력하세요.
```py
scores = [88, 92, 79, 95,
          75, 68, 82, 70,
          92, 99, 95, 88,
          60, 72, 65, 58,
          85, 80, 78, 90]

````python
import numpy as np
````

1. 위 리스트를 5행 4열 ndarray로 변환하고 shape, ndim, size, dtype을 출력하시오.

````python
# 문제 1-1번
scores = [88, 92, 79, 95,
          75, 68, 82, 70,
          92, 99, 95, 88,
          60, 72, 65, 58,
          85, 80, 78, 90]


x = np.array(scores).reshape(5, 4)
print(x.shape, x.ndim, x.size, x.dtype)
````

2. 학생 별 평균과 과목 별 평균을 각각 구하시오.

````python
# 문제 1-2번
student_means = np.mean(x, axis=1)
subject_means = np.mean(x, axis=0)
print(student_means)
print(subject_means)
````

3. 전체에서 가장 높은 점수와 그 점수가 몇 번째 학생·몇 번째 과목인지 구하시오.

````python
# 문제 1-3번
print(x.max()) # 최대값
index_max = np.where(x == x.max()) # 최대값 위치
result = int(index_max[0][0]), int(index_max[1][0])
print(result)
````

4. 전체 평균, 표준편차, 분산을 구하시오.

````python
# 문제 1-4번
print(np.mean(x)) # 평균
print(np.std(x)) # 표준편차
print(np.var(x)) # 분산
````

5. 평균이 80점 이상인 학생의 인덱스와 인원수를 구하고, 해당 학생들의 성적만 잘라내어 출력하시오.

````python
# 문제 1-5번
index_student = np.where(student_means >= 80)[0] # 조건에 맞는 위치
count = np.sum(student_means >= 80)

print(index_student)
print(count)
print(x[student_means>=80])
````

6. 90점 이상인 점수의 개수, 100점 이상인 점수가 하나라도 있는지, 모든 점수가 60점 이상인지를 판별하시오.

````python
# 문제 1-6번
count = np.sum(x >= 90)

print(count) # 90점 이상인 갯수
print(np.any(x >= 100)) # 하나라도 참이어야 참
print(np.all(x >= 60)) # 모든 요소가 참이어야 참
````

7. 90점 이상은 'A', 80점 이상은 'B', 나머지는 'C'로 변환한 등급 배열을 만드시오.

````python
# 문제 1-7번
grade = np.where(x >= 90, 'A', np.where(x >= 80, 'B', 'c'))  
print(grade)
````

import numpy as np
np.random.seed(2024)
img = np.random.randint(0, 256, size=(8, 8))

## 3. 지점별 판매 데이터에 브로드캐스트와 행렬 연산을 적용해 매출을 분석하시오
3개 지점이 4개 제품을 판매한 수량과, 제품별 단가가 다음과 같습니다.
```py
sales = np.array([[120,  80,  45,  60],     # A지점
                  [200, 150,  90,  30],     # B지점
                  [ 75,  60, 120,  95]])    # C지점
price = np.array([1500, 3000, 5000, 2000])  # 제품 1~4 단가

````python
sales = np.array([[120,  80,  45,  60],     # A지점
                  [200, 150,  90,  30],     # B지점
                  [ 75,  60, 120,  95]])    # C지점      # 3 x 4 배열
price = np.array([1500, 3000, 5000, 2000])  # 제품 1~4 단가    # 4 X 1 배열
````

1. 브로드캐스트를 이용해 각 지점의 제품별 매출액(수량 × 단가) 행렬을 구하시오. (반복문 금지)

````python
total = sales * price  # 2차원 * 1차원
print(total)
````

2. 행렬곱을 이용해 각 지점의 총매출을 한 번에 구하시오. 또한 제품별 총판매량도 구하시오.

````python
# 행렬곱을 이용한 각 지점의 총 매출
print(sales @ price)

# 제품별 총 판매량
print(sales.sum(axis=0))
````

3. 각 제품의 평균 판매량을 구한 뒤, 브로드캐스트로 평균 대비 편차 행렬을 구하시오.

````python
# 제품별 평균 판매량
sales.mean(axis=0)

# 평균 대비 편차 행렬
print(sales - sales.mean(axis=0))
````

4. 각 지점을 기준으로 판매량을 정규화하시오. 결과의 행별 평균이 0, 표준편차가 1이 되는지 확인하시오.

````python
# z-점수 정규화
z = (sales - sales.mean(axis=1).reshape(-1,1)) / sales.std(axis=1).reshape(-1,1)    # (값 - 평균) / 표준편차
print(np.round(z, 2))
````

5. sales의 전치행렬을 구하고 shape을 출력하시오.

````python
# sales의 전치행렬(행, 렬을 바꾼 것)
# shape 출력
sales.T.shape
````

6. sales의 앞 3개 열만 잘라 3×3 정방행렬 A를 만들고, 행렬식과 역행렬을 구하시오. 이어서 A와 역행렬을 곱해 단위행렬이 나오는지 확인하시오.

````python
A = sales[:, :3]

print(np.linalg.det(A))   # 행렬식
print(np.linalg.inv(A))   # 역행렬

print(A @ np.linalg.inv(A))   # A와 역행렬 곱
````

7. 총매출이 가장 높은 지점의 인덱스와, 지점 순서대로의 누적 매출을 구하시오.

````python
# 총매출이 가장 높은 지점 인덱스
total_sales = sales @ price
print(np.argmax(total_sales))   # B지점

# 지점 순서대로의 누적 매출
print(total_sales.cumsum())
````
