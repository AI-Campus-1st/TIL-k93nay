```python
아래와 같은 scores.csv가 주어집니다 (일부 행에 빈 값 또는 숫자가 아닌 값이 섞여 있습니다).

name,kor,eng,math
홍길동,90,85,80
김철수,,70,60
이영희,88,abc,95
박민수,100,100,100
각 학생의 평균 점수를 계산해 result.csv(name,average)로 저장하는 프로그램을 작성하세요.

점수 칸이 비어 있거나 숫자로 변환할 수 없으면 그 과목은 0점으로 처리한다.

변환에 실패한 칸이 있으면 "[처리] 이영희 - math 값 오류(abc) → 0점 처리" 형태로 로그를 출력한다.

평균은 소수 첫째 자리까지 반올림한다.

실행 결과

[처리] 김철수 - kor 값 오류() → 0점 처리
[처리] 이영희 - eng 값 오류(abc) → 0점 처리
완료: result.csv 생성
```

```python
with open('scores.csv', 'r', encoding='utf-8') as f:
    columns = f.readline().rstrip().split(',')  # ['name', 'kor', 'eng', 'math']
    lines = f.readlines()

print(columns)
print(lines)
```

```python
datas = []
subject_names = columns     # 목록이름   # ['name', 'kor', 'eng', 'math']

for line in lines:
    data = line.rstrip().split(',')    #[['홍길동','90','85','80']] 형태로 분리
    name = data[0]          # 학생이름

    for i in range(1,4):    # 과목 점수 있는 부분
        subject_score = data[i] if i < len(data) else ""   # 데이터가 없는 점수 부분은 ""(빈 문자열)으로 남겨두는 것   

        try:
            data[i] = int(subject_score)      # 점수 부분을 str -> int로 바꿈

        except ValueError:
            print(f'[처리] {name} - {subject_names[i]} 값 오류({subject_score}) -> 0점 처리')
            data[i] = 0

        # 김철수 kor점수부분 처리단계
        # 1. ['김철수', '', '70', '60']에서 kor 점수 부분이 없으므로 빈 문자열로 처리
        # 2. try의 if문 시도 -> 오류발생(ValueError) -> except문으로 이동
        # 3. except문에서 "[처리] 김철수 - kor 값 오류() → 0점 처리" 출력 후 data[i] = 0으로 처리

        # 이영희 eng 점수부분 처리단계
        # 1. ['이영희', '88', 'abc', '95']에서 eng 점수 부분이 문자열이지만, 데이터가 있으므로 그대로 처리
        # 2. try의 if문 시도 -> 오류발생(ValueError) -> except문으로 이동
        # 3. except문에서 "[처리] 이영희 - eng 값 오류() → 0점 처리" 출력 후 data[i] = 0으로 처리

    datas.append(data)   # for line in lines 내 반복내용(이름포함 리스트에 업데이트)이므로 들여쓰기로 진행
```

```python
print(datas)
```

```python
with open('result.csv', 'w') as out:
    out.write("name, average\n")   # 한 줄에 name, average를 기록

    for row in datas:                  # 그 다음줄부터 각 학생의 이름과 평균을 기록
        name = row[0]
        total = row[1] + row[2] + row[3]  # kor, eng, math 점수만 합산
        avg = round(total / 3, 1)  # 과목 수(3)로 나누고 소수 첫째 자리까지 반올림

        out.write(f"{name},{avg}\n") # 파일에 기록

print('완료: result.csv 생성')
```

```python
CSV를 읽어 JSON으로 변환하되, 실패한 행은 건너뛰고 오류 리포트를 남긴다.

members.csv를 읽어 members.json(리스트 형태)으로 변환하는 프로그램을 작성하세요.
```

```python
f = open('members.csv', 'r', encoding='utf-8')
columns = f.readline().rstrip().split(',')  # ['id', 'name', 'age']
lines = f.readlines()
print(columns)
print(lines)
```

```python
id,name,age
1,홍길동,30
2,김철수,스물다섯
3,이영희,28
4,,40
```

```python
규칙:

각 행을 {"id": 정수, "name": 문자열, "age": 정수} 형태의 딕셔너리로 변환한다.
```

```python
datas = []

for line in lines:
    data = line.rstrip().split(',')  # ['1', '홍길동', '30'] 형태로 분리
    
    for i in range(len(data)):        
        try:
            data[i] = int(data[i])    # id, age 정수화

        except ValueError:
            pass
        
    temp = dict(zip(columns, data))
    datas.append(temp)


f.close()
```

```python
datas
```

```python
규칙:

age가 숫자가 아니거나 name이 비어 있는 행은 JSON에 넣지 않고 건너뛴다.
```

```python
f = open('members.json', 'w', encoding='utf-8')


filtered_datas = []                
for data in datas:
    if not isinstance(data['age'], int) or data['name'] == "":
        continue        # 건너뛰기
    filtered_datas.append(data)

import json
json.dump(filtered_datas, f, ensure_ascii=False, indent=2)

f.close()
```

```python
건너뛴 행은 errors.log 파일에 행번호와 사유를 기록한다(예: 3행: name 비어있음).

최종적으로 "성공 N건 / 실패 M건"을 화면에 출력한다.

정상 변환된 데이터만 members.json에 한글이 깨지지 않게 저장한다.
```

```python
with open('errors.log', 'w', encoding='utf-8') as error_file:

    for i, data in enumerate(datas):

        if data['name'] == "":          # name이 비어있는 경우
            error_file.write(f"{(i+2)}행: name 비어있음\n")
            continue

        # name이 비어있지 않은 경우, age가 숫자가 아닌 경우
        try:
            data['age'] = int(data['age'])  # age를 int로 변환 시도. 성공하면 error_file에 기록하지 않음

        except ValueError as err:
            error_file.write(f"{(i+2)}행: {err}\n")

print(f"성공 {len(filtered_datas)}건 / 실패 {(len(datas) - len(filtered_datas))}건")
```
