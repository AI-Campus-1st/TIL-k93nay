```python
import matplotlib.pyplot as plt
from sklearn.datasets import make_regression
```

```python
# 데이터 생성
data = make_regression(n_samples=1000, n_features=1, bias=200, noise=30, random_state=42) 

# 기본 파이썬 리스트 자료형으로 변환
X_list = [float(val[0]) for val in data[0]]
y_list = [float(val) for val in data[1]]

n_data = len(X_list)

n_data
```

```python
# 가중치(w1) 및 편향(w0) 초기화
w1 = 0.0
w0 = 0.0

# 하이퍼파라미터 설정 (에포크 : 총 반복 횟수/ 학습률 : 한번 에포크마다 업데이트할 규모(폭))
epochs = 1000
learning_rate = 0.01

# 학습 시작
for epoch in range(epochs):
    cost = 0.0
    dw1 = 0.0
    dw0 = 0.0
    
    # 각 데이터 포인트를 순회하며 오차 및 기울기 계산
    for i in range(n_data):
        x_i = X_list[i]
        y_i = y_list[i]
        
        # 가설 함수 계산 (y = w1 * x + w0)
        y_pred = w1 * x_i + w0
        
        # 손실(MSE) 누적 계산
        cost += (y_i - y_pred) ** 2
        
        # 편미분 값(기울기) 누적 계산
        dw1 += (y_pred - y_i) * x_i  # 미분공식에서 *2와 /n을 제외한 공식
        dw0 += (y_pred - y_i)
        
    # 평균 계산 (MSE 미분값)
    cost /= n_data
    dw1 = (dw1 * 2) / n_data   # 미분공식 완성
    dw0 = (dw0 * 2) / n_data
    
    # 가중치 업데이트(경사하강법. 기울기의 반대방향으로 가중치 업데이트이므로 -)
    w1 -= learning_rate * dw1
    w0 -= learning_rate * dw0
    
    # 100회마다 학습 상태 출력
    if (epoch + 1) % 100 == 0:
        print(f"Epoch {epoch+1:4d} | Cost: {cost:.4f} | w1: {w1:.4f} | w0: {w0:.4f}")

print("\n--- 기본 파이썬 구현 최종 결과 ---")
print(f"가중치(w1): {w1:.4f}, 편향(w0): {w0:.4f}")
```

# 문항 2 : 다변량 선형회귀 모델 직접 구현

## 요구사항

독립 변수의 개수가 100개인 선형회귀 모델을 기본 파이썬 자료형만 활용하여 직접 구현해보세요.

numpy 라이브러리를 활용하여 구현해보세요.

from sklearn.datasets import make_regression

data = make_regression(n_samples = 5000, n_features = 100, bias = 200, noise = 50) 

X = data[0]
y = data[1]

```python
# 데이터 생성
data = make_regression(n_samples=5000, n_features=100, bias=200, noise=50, random_state=42) 

# 순수 파이썬 리스트 자료형으로 변환
X_list = [list(row) for row in data[0]]
y_list = [float(val) for val in data[1]]

n_data = len(X_list)        # 5000
n_features = len(X_list[0])    # 100
```

```python
# 100개의 가중치(w) 리스트와 1개의 편향(w0) 초기화 (모두 0.0으로 시작)
w = [0.0] * n_features
w0 = 0.0

# 하이퍼파라미터 설정
epochs = 500
learning_rate = 0.05  # 특성이 많을 때는 학습률과 에포크를 적절히 조절합니다.

# 학습 시작
for epoch in range(epochs):
    cost = 0.0
    dw = [0.0] * n_features  # 100개 가중치의 기울기를 담을 리스트
    dw0 = 0.0
    
    # 5000개의 데이터를 하나씩 순회
    for i in range(n_data):
        x_i = X_list[i]
        y_i = y_list[i]
        
        # 가설 함수 계산: y_pred = (w1*x1 + w2*x2 + ... + w100*x100) + w0
        y_pred = 0.0
        for j in range(n_features):
            y_pred += w[j] * x_i[j]
        y_pred += w0
        
        # 손실 누적 (MSE용)
        error = y_pred - y_i
        cost += error ** 2
        
        # 기울기 누적 계산
        for j in range(n_features):
            dw[j] += error * x_i[j]
        dw0 += error
        
    # 평균 계산 및 미분 상수 반영
    cost /= n_data
    dw0 = (dw0 * 2) / n_data
    for j in range(n_features):
        dw[j] = (dw[j] * 2) / n_data
        
    # 가중치 및 편향 업데이트
    w0 -= learning_rate * dw0
    for j in range(n_features):
        w[j] -= learning_rate * dw[j]
        
    # 100회마다 손실과 일부 가중치, 편향 상태 출력
    if (epoch + 1) % 100 == 0:
        print(f"Epoch {epoch+1:3d} | Cost: {cost:.4f} | w0(편향): {w0:.4f} | w[0](첫번째 가중치): {w[0]:.4f}")

print("\n--- 기본 파이썬 구현 최종 결과 ---")
print(f"편향(w0): {w0:.4f}")
print(f"앞의 5개 가중치(w): {[round(val, 4) for val in w[:5]]} ...")
```

```python
# 데이터 생성 및 Numpy 배열 준비
data = make_regression(n_samples=5000, n_features=100, bias=200, noise=50, random_state=42) 
x = data[0]               # Shape: (5000, 100)
y = data[1].reshape(-1, 1) # Shape: (5000, 1)

n_data = len(x)
```

```python
# 가중치(100, 1) 및 편향(1) 초기화
w = np.zeros((100, 1))
w0 = 0.0

# 하이퍼파라미터 설정
epochs = 500
learning_rate = 0.05

history = {'cost': []}

# 학습 시작
for epoch in range(epochs):
    # 예측값 계산 (5000, 100) @ (100, 1) + w0 = (5000, 1)
    y_pred = x @ w + w0
    
    # 비용(MSE) 계산
    cost = np.sum((y_pred - y) ** 2) / n_data
    history['cost'].append(cost)
    
    # 기울기 계산 및 가중치 업데이트
    # (5000, 1).T @ (5000, 100) -> (1, 5000) @ (5000, 100) = (1, 100) -> 다시 세우면 (.T) -> (100, 1)
    dw = (2 * (y_pred - y).T @ x).T / n_data
    dw0 = 2 * np.mean(y_pred - y)
    
    w -= learning_rate * dw
    w0 -= learning_rate * dw0
    
    # 100회마다 학습 상태 출력
    if (epoch + 1) % 100 == 0:
        print(f"Epoch {epoch+1:3d} | Cost: {cost:.4f} | w0(편향): {w0:.4f} | w[0](첫번째 가중치): {w[0][0]:.4f}")

print("\n--- Numpy 구현 최종 결과 ---")
print(f"편향(w0): {w0:.4f}")
print(f"앞의 5개 가중치(w): {w[:5].ravel()}")
```
