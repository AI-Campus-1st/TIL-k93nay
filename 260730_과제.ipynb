```python
class Animal:
	def __init__(self, name):
		self.name = name
	
	def walk(self):
		print(f'{self.name}! 걷는다!')
	
	def eat(self):
		print(f'{self.name}! 먹는다!')
```

```python
class Dog(Animal):

    def __init__(self, name):
        super().__init__(name)

    def walk(self):
        print(f'{self.name}! 달린다!')

    def bark(self):
        print(f'{self.name}! 짖는다!')

class Bird(Animal):

    def __init__(self, name):
        super().__init__(name)

    def fly(self):
        print(f'{self.name}! 푸드덕!')
```

```python
dog = Dog('멍멍이')
dog.walk() # 멍멍이! 달린다!
dog.bark() # 멍멍이! 짖는다!

bird = Bird('구구')
bird.walk() # 구구! 걷는다!
bird.eat() # 구구! 먹는다!
bird.fly() # 구구! 푸드덕!
```

```python
from abc import ABC, abstractmethod      # 추상 매서드 가져와야함

class Payment(ABC):                      # ABC 클래스를 상속
    def __init__(self, owner, balance):
        self.owner = owner
        self.balance = balance

    @abstractmethod                      # 추상 매서드 적용
    def fee(self, amount):
        """
        결제금액에 대한 수수료를 반환
        """

    def pay(self, amount):
        current_fee = self.fee(amount)          # 최종수수료 값에 대한 정의
        total_amount = amount + current_fee     # 최종 금액에 대한 정의
        if self.balance < total_amount:
            raise ValueError("잔액 부족")
        else:                                   # 잔액이 최종 금액보다 클 경우
            self.balance = self.balance - total_amount
            return {                             # 딕셔너리 반환
                'amount': amount, 
                'fee': current_fee, 
                'balance': self.balance
                }
```

```python
class CardPayment(Payment):
    def fee(self, amount):                       # 부모 클래스에 있던 추상 매서드 구체화
        current_fee = int(amount*0.025)          # 부모 클래스 pay매서드에서 self.fee 부분에서 자식 매서드로 넘어옴
        print(f'수수료 {current_fee}')
        return current_fee                       # 계산된 값 부모 클래스의 pay매서드에 적용(=current fee에 넣음)

class MobilePayment(Payment):
    def fee(self, amount):
        print(f'수수료 100')
        return 100                               # 계산된 값 부모 클래스의 pay매서드에 적용(=current fee에 넣음)

class PointPayment(Payment):
    def fee(self, amount):
        print(f'수수료 0')
        return 0                                 # 계산된 값 부모 클래스의 pay매서드에 적용(=current fee에 넣음)
        
    def pay(self, amount):
        result = super().pay(amount)
        result["point_used"] = True              # 딕셔너리에 key하나 추가하는 것(value포함)
        return result
```

```python
# 실행 코드
print(CardPayment("홍길동", 20000).pay(10000))    # 수수료 250
print(MobilePayment("김철수", 20000).pay(10000))  # 수수료 100
print(PointPayment("이영희", 20000).pay(10000))   # 수수료 0, point_used True

print("---")
try:
    CardPayment("최가난", 5000).pay(10000)         # 잔액 부족
except ValueError as e:
    print(f"결제 실패: {e}")
```
