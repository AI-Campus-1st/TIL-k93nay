```python
import time
from konlpy.tag import Kkma

kkma = Kkma()

stopPos = ['J', 'S', 'NR', 'OL', 'X']
stopwords = [
    '기획', '10월', '올영픽', '단품', '용량', '대용량',
    '/', '[', ']', ',', 'ml', '+', '(', ')', '더블', '증정'
    ]

unique_tokens = set()

for doc, y in training_set:
    raw_tokens = kkma.pos(doc)
    for word, pos in raw_tokens:
        if word in stopwords or pos in stopPos or len(word) == 1:
            continue
        unique_tokens.add(word)     # 단어 목록 정리
```

```python
import pandas as pd

len_y = 3  # 스킨케어(0), 메이크업(1), 헤어케어(2)
temp = {t: [0] * len_y for t in unique_tokens}

# training_set을 순회하며 카테고리별 단어 빈도수(Count) 누적 계산
for doc, y in training_set:
    raw_tokens = kkma.pos(doc)
    for word, pos in raw_tokens:
        if word in stopwords or pos in stopPos or len(word) == 1:
            continue
        
        if word in temp:
            temp[word][y] += 1

# 판다스 데이터프레임으로 변환 후 형태 확정 (.T)
df = pd.DataFrame(temp, index=['skincare', 'makeup', 'haircare']).T

print("\n=== 2단계: 카테고리별 형태소 빈도수 완성 표 ===")
display(df.head(15))
```

```python
targets = df.columns  

k = 0.5
for target in targets:     
    column_name = f'P(w|{target})'
    df[column_name] = (k + df[target]) / (2 * k + df[target].sum())  

import numpy as np
for target in targets:
    conditional_column = f'P(w|{target})'
    log_column = f'Log(P(w|{target}))'
    df[log_column] = np.log(df[conditional_column]) 


df
```

```python
for target in targets:
    log_p_targets = np.log(df[targets].sum() / df[targets].sum().sum()).to_dict()
    
log_p_targets
```

```python
import numpy as np

# 테스트할 새로운 상품명
new_doc = "에스트라 아토베리어 크림 앰플 기획 세트"

# 각 카테고리별 초기 점수를 사전 확률 로그값으로 세팅
result = {target: log_p_targets[target] for target in targets}

# 새로운 문장을 토큰화하여 각 단어의 로그 조건부 확률을 누적 더하기
new_tokens = kkma.pos(new_doc)
for word, pos in new_tokens:
    if word in stopwords or pos in stopPos or len(word) == 1:
        continue
    
    for target in targets:
        log_column = f'Log(P(w|{target}))'
        result[target] += df[log_column].get(word, 0.0)

# 비율 계산
max_log = max(result.values()) # 로그를 씌우면 -값인데, 언더플로우 방지를 위해 최댓값(0에 가까운값으로) 스케일링   
exp_scores = {target: np.exp(score - max_log) for target, score in result.items()}
sum_exp = sum(exp_scores.values())

final_probabilities = {target: score / sum_exp for target, score in exp_scores.items()}

# 최종 예측 결과 출력
print(f"입력한 상품명: '{new_doc}'")
print("-" * 50)
for target, prob in final_probabilities.items():
    print(f"[{target}] 카테고리일 확률: {prob * 100:.2f}%")
print("-" * 50)

best_target = max(final_probabilities, key=final_probabilities.get)
print(f"최종 예측 결과: 이 상품은 [{best_target}] 카테고리입니다!")
```
