# 문항 1 : 할리스커피 매장 정보 10페이지 수집

### 할리스커피 매장 찾기 페이지에서 매장 정보를 수집해 리스트-딕셔너리 형태로 정리하시오.

대상: https://www.hollys.co.kr/store/korea/korStore2.do

총 10페이지를 순회할 것 (페이지를 넘기며 URL이 바뀌는 것을 확인)

추출 필드: 지역 / 매장명 / 현황 / 주소 / 매장 서비스 / 전화번호

매장 서비스는 리스트로 담을 것 (아이콘이 여러 개인 매장이 있음)

결과를 hollys.csv로 저장할 것

### 결과 예시
[{'지역': '서울 동대문구',
  '매장명': '경희대 경영대점',
  '현황': '영업중',
  '주소': '서울특별시 동대문구 경희대로 26 (회기동) 경영대학 3층',
  '매장 서비스': ['주차'],
  '전화번호': '.'},
 ...]

```python
import requests
from bs4 import BeautifulSoup
import time
```

```python
base_url = "https://www.hollys.co.kr/store/korea/korStore2.do"

hollys_stores = []
for page in range(1, 11):
    url = f"{base_url}?pageNo={page}"
    
    response = requests.get(url)
    
    # 정상적으로 가져왔는지 상태 확인
    if response.status_code == 200:
        print(f"현재 {page}페이지 요청 성공!")
        soup = BeautifulSoup(response.text, 'html.parser')

        trs = soup.select('table.tb_store tbody tr')  # tb_store class 내 tbody안에 있는 tr(행 전체)
        for tr in trs:
            tds = tr.select('td')    # 하나의 행 안에 있는 모든 td를 리스트로 가져옴
            if len(tds) < 6:     # 6가지보다 적은 정보가 있는 행 혹은 빈 행은 패스
                continue

            region = tds[0].text.strip()       # 지역
            name = tds[1].text.strip()         # 매장명
            status = tds[2].text.strip()       # 현황
            address = tds[3].text.strip()      # 주소
            tel = tds[5].text.strip()          # 전화번호

            services = []                          # 매장 서비스
            icons = tds[4].select('img')           # 이미지(아이콘) 등 리스트로 가져옴
            for img in icons:
                if img.get('alt'):                     # alt : 대체 텍스트
                    services.append(img.get('alt'))    # 대체 텍스트로 services 리스트에 넣음

            store_info = {
                '지역': region,
                '매장명': name,
                '현황': status,
                '주소': address,
                '매장 서비스': services,
                '전화번호': tel
                }

            hollys_stores.append(store_info)     # 최종 가게 정보 넣기
        
        time.sleep(1)
    else:
        print(f"{page}페이지 요청 실패. 상태 코드: {response.status_code}")
```

```python
hollys_stores
```

```python
import csv
```

```python
with open('hollys.csv', 'w', encoding='utf-8-sig', newline='') as f:
    
    fieldnames = ['지역', '매장명', '현황', '주소', '매장 서비스', '전화번호']
    
    writer = csv.DictWriter(f, fieldnames=fieldnames)        # 딕셔너리 전용. 키값을 기준으로 열 매칭
    
    writer.writeheader()                                     # 맨 첫줄에 fieldnames(열 제목) 작성
    
    writer.writerows(hollys_stores)                          # 매장정보(리스트 내 딕셔너리를 한 행씩) 쓰기
```

# 문항 2 : 알라딘 베스트셀러 수집

## 알라딘 베스트셀러 페이지에서 도서 정보를 수집해 CSV로 저장하시오.

대상: 알라딘 베스트셀러 목록 페이지(https://www.aladin.co.kr/shop/common/wbest.aspx?BranchType=1)

추출 필드: 카테고리 / 제목 / 저자 / 할인가격 / 이미지 URL

이미지 URL은 <img> 태그의 속성에서 가져올 것

결과를 aladin_bestseller.csv로 저장할 것

추가학습: 총 500위까지 수집해주세요.

## 결과 예시 (8월 2주)

[{'카테고리': '[국내도서]',
  '제목': '오뒷세이아',
  '저자': '호메로스',
  '정가': '25,000원',
  '할인가격': '22,500원',
  '이미지': '<https://image.aladin.co.kr/product/39940/12/cover200/8932476462_2.jpg>'},
 ...]

```python
import re
```

```python
base_url = "https://www.aladin.co.kr/shop/common/wbest.aspx"

best_sellers = []
for page in range(1, 11):
    url = f"{base_url}?BestType=Bestseller&BranchType=1&CID=0&page={page}"
    
    headers = {
        'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/120.0.0.0 Safari/537.36'
    }
    response = requests.get(url, headers=headers)
    
    if response.status_code == 200:
        print(f"현재 {page}페이지 요청 성공!")
        soup = BeautifulSoup(response.text, 'html.parser')
        
        book_lists = soup.select('div.ss_book_list')
        for box in book_lists:
            lis = box.select('ul li')
            
            category_el = box.select_one('span.tit_category')
            if not category_el:
                continue
                
            category = category_el.text.strip()                   # 카테고리
            title = box.select_one('a.bo3').text.strip()          # 제목

            info_li = category_el.find_parent('li')
            author_li = info_li.find_next_sibling('li')
            author_raw = author_li.text.split('|')[0].strip()
            first_author = author_raw.split(',')[0].strip()
            author = re.sub(r'\(.*?\)', '', first_author).strip()    # 저자명

            price_el = box.select_one('span.ss_p2')
            sale_price = price_el.text.strip() if price_el else "0원"   # 할인가격

            price_li = price_el.find_parent('li') if price_el else None
            if price_li:
                price_text = price_li.text.strip()
                if '→' in price_text:
                    regular_price = price_text.split('→')[0].strip()
                    if '원' not in regular_price: 
                        regular_price += '원'                        # 정가
                else:
                    regular_price = sale_price
            else:
                regular_price = "0원"
            
            img_tag = box.find_parent('table').select_one('img')
            img_url = img_tag.get('src') if img_tag else ""       # 이미지 URL
            
            book_info = {
                '카테고리': category,
                '제목': title,
                '저자': author,
                '정가': regular_price,
                '할인가격': sale_price,
                '이미지 URL': img_url
            }
            
            
            best_sellers.append(book_info)
        
        time.sleep(1) 
    else:
        print(f"{page}페이지 요청 실패. 상태코드: {response.status_code}") 
```

```python
with open('aladin_bestseller.csv', 'w', encoding='utf-8-sig', newline='') as ff:
    
    fieldnames = ['카테고리', '제목', '저자', '정가', '할인가격', '이미지 URL']
    
    writer = csv.DictWriter(ff, fieldnames=fieldnames)        # 딕셔너리 전용. 키값을 기준으로 열 매칭
    
    writer.writeheader()                                     # 맨 첫줄에 fieldnames(열 제목) 작성
    
    writer.writerows(best_sellers)                          # 매장정보(리스트 내 딕셔너리를 한 행씩) 쓰기
```
